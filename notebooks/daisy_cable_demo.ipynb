# DAiSy Cable Rehang Demo

This notebook launches the DAiSy cable rehanging demonstration from
`cognitive_robot_abstract_machine/coraplex/demos/daisy_cable_demo`.

The plan runs in **simulation**: the DAiSy robot (two UR5 arms), a cable hanger and a
cable are built from the DAiSy URDF, and the grasp / regrasp / rehang motions are
executed by Giskard without any real hardware.


## Setup

1. **Kernel** — make sure the top-right of this notebook says **CRAM**. If not, use
   *Kernel → Change Kernel… → CRAM*.
2. **Visualisation** — the RViz robot viewer opens in a sidecar panel to the
   side (it starts automatically when the container boots, so it’s ready before
   the demo runs). You can also open it full-screen via *Launcher → RViz*.


```python
import os
from IPython.display import IFrame
from sidecar import Sidecar

sidecar = Sidecar(title="RViz", anchor="split-right")

base_url = os.environ.get("JUPYTERHUB_SERVICE_PREFIX", "/").rstrip("/")
rviz_url = f"{base_url}/rviz/"

with sidecar:
    display(IFrame(src=rviz_url, width="100%", height="720px"))
```

```python
import os
import sys
from pathlib import Path

# The demo lives inside the cognitive_robot_abstract_machine checkout. Add its
# directory to the path so its local imports (define_sim_daisy, ...) resolve.
repo_dir = Path(os.environ.get("REPO_DIR", Path.home() / "vrb-cable-demo"))
demo_dir = (
    repo_dir
    / "cognitive_robot_abstract_machine"
    / "coraplex"
    / "demos"
    / "daisy_cable_demo"
)
sys.path.insert(0, str(demo_dir))

from demo import DAiSyCableDemo
from coraplex.datastructures.enums import ExecutionType

demo = DAiSyCableDemo()
demo.load_world(execution_mode=ExecutionType.SIMULATED)
demo.load_environment()

print("DAiSy cable demo ready. Click the button below to launch it.")
```

```python
import ipywidgets as widgets
from IPython.display import display

launch_button = widgets.Button(description="Launch Demo", button_style="success")
output = widgets.Output()

def launch(_button):
    launch_button.disabled = True
    with output:
        output.clear_output()
        print("Launching the DAiSy cable demo in simulation...")
        demo.load_environment()
        demo.run_demo(execution_mode=ExecutionType.SIMULATED)
        print("Demo finished.")

launch_button.on_click(launch)
display(widgets.VBox([launch_button, output]))
```

> **Note** — the demo runs in this kernel, so the notebook is busy while the plan
> executes. To run it again, restart the kernel (*Kernel → Restart Kernel*) and
> re-run the cells above.
